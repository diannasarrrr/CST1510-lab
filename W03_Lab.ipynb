### C2. How many are over?

Write `count_over(*values, limit=100)` with a one-line docstring. It returns how many of the values
are over the limit.

| Call | Expected output |
|---|---|
| `count_over(87, 120, 45, 101)` | `2` |
| `count_over(87, 120, 45, 101, limit=50)` | `3` |
| `count_over()` | `0` |

````python
# C2
def count_over(*values, limit=100):
    count = 0   
    for v in values:
        if v > limit:
            count += 1
    return count
print(count_over(87, 120, 45, 101))
print(count_over(87, 120, 45, 101, limit=50))
print(count_over())
````

## What to build

**Threshold — pass standard.** Write `status_of(percent)`. Ask for the three values,
work out the difference and percentage inline, call `status_of` for the status, and
print a bordered report including it.

```
==================================
```

**Typical.** Also write `check(value, limit)`, returning the difference and

**Excellent.** Also write `print_report(...)` — every `print()` for the report
happens inside it, nowhere else. Wrap the input/process/output steps in a loop so you
can check as many records as you like in one run — type `quit` to stop. Keep count of
how many came back `OVER LIMIT` and print that once, after the loop ends.

## Rules

- **Every function does one job.** If a function both calculates and prints, split it.
- Do not calculate the same thing in two different places.
- **Give every function you write a one-line docstring** saying what it does.
- **No lists yet** — that is Week 4.

## When it works

1. Run it three times with different inputs. Does it still look right?
2. Run it with a total of `0`. Write the error down.

````python
"""
RECORD CHECK  -  Cyber Security
===============================

Name  : Dian Nasar
Lane  : Cyber Security
Date  :  07 October 2026

Run it:   python template.py
"""

# =================================================================== FUNCTIONS


# 1. Function to evaluate status threshold
def status_of(percent):
    """Return a status string based on the given failed login percentage."""
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= 90:
        return "WARNING"
    else:
        return "OK"


# 1. Function to calculate difference and percentage
def check(value, limit):
    """Calculate and return remaining allowable attempts and failed login percentage."""
    difference = limit - value
    percent = (value / limit) * 100
    return difference, percent


# 1. Function to handle all printing output
def print_report(label, value, limit, difference, percent, status):
    """Print the formatted bordered report for a security record check."""
    print()
    print("=" * 34)
    print(f"  RECORD CHECK  -  {label}")
    print("=" * 34)
    print(f"  Failed Logins : {value:>10.2f}")
    print(f"  Total Attempts: {limit:>10.2f}")
    print(f"  Attempts Left : {difference:>10.2f}")
    print(f"  Percent Failed: {percent:>10.2f} %")
    print(f"  Status        : {status:>12}")
    print("=" * 34)


# ========================================================== MAIN EXECUTION LOOP

over_limit_count = 0

while True:
    # 2. Get user input
    label = input("\nEnter Source IP (or 'quit' to exit): ").strip()
    if label.lower() == "quit":
        break

    value = float(input("Enter failed logins count: "))
    limit = float(input("Enter total login attempts: "))

    # 3. Process inputs using functions
    difference, percent = check(value, limit)
    status = status_of(percent)

    if status == "OVER LIMIT":
        over_limit_count += 1

    # 4. Output the report
    print_report(label, value, limit, difference, percent, status)

print(f"\nSession finished. Total OVER LIMIT records: {over_limit_count}")
````
